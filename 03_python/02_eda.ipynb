# Convert Dates


```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

# Create Time Features


```python
df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()
df["quarter"] = df["order_date"].dt.quarter
df["day_name"] = df["order_date"].dt.day_name()
```

# Sales Distribution

### Objective: 
Understand the distribution of sales across all transactions.

### Why this analysis?
- Detect skewness
- Identify outliers
- Understand customer spending behaviour

## Sales Distribution


```python
plt.figure(figsize=(10,5))

sns.histplot(df["sales_amount"], bins=30, kde=True)

plt.title("Sales Distribution")

plt.show()
```

##  Observations

- Most orders have relatively low sales values.
- A few high value orders create a right-skewed distribution.
- High value transactions are comparatively rare.

---------------------------------------------------------------------------------------------------

# Sales by Product Category

### Objective
 Compare total sales generated by each product category.

### Visualization
 Bar Chart

# Sales by Category

```python
category_sales = (
    df.groupby("product_category")["sales_amount"]
      .sum()
      .sort_values(ascending=False)
)

category_sales.plot(kind="bar", figsize=(10,5))

plt.ylabel("Sales")

plt.show()
```

## Observations
- One category contributes the highest share of total sales that is "Electronics".
- Some categories generate significantly lower revenue like "Furniture", "Sports" and others.
- Product categories contribute unevenly to overall business performance.
- High-performing categories deserve greater marketing attention.

------------------------------------------------------------------------------------------------------------------------------------------------------------

# Profit by Product Category

### Objective
 Evaluate profitability across product categories.

### Visualization
 Bar Chart

# Profit by Category

```python
profit = (
    df.groupby("product_category")["profit"]
      .sum()
      .sort_values(ascending=False)
)

profit.plot(kind="bar", figsize=(10,5))

plt.ylabel("Profit")

plt.show()
```

## Observations
- High sales do not always translate into high profit.
- Certain categories achieve stronger profit margins.
- Some categories require pricing or discount optimisation.

------------------------------------------------------------------------------------------------------------------------------------------------------------

# Top 10 Customers by Sales

# Monthly Sales Trend

### Objective
 Analyse sales performance over time.

### Visualization
 Line Chart

# Monthly Sales Trend

```python
monthly_sales = (
    df.groupby(["year","month"])["sales_amount"]
      .sum()
      .reset_index()
)

plt.figure(figsize=(12,5))

plt.plot(monthly_sales["sales_amount"])

plt.title("Monthly Sales Trend")

plt.xlabel("Timeline (Year-Month)", fontsize=11, fontweight="bold")

plt.ylabel("Total Sales Amount ($)", fontsize=11, fontweight="bold")


plt.show()
```

## Observations
- Sales fluctuate across different months.
- Seasonal patterns may exist.
- Growth periods and declining periods are clearly visible.
- Trend analysis supports future sales forecasting.

------------------------------------------------------------------------------------------------------------------------------------------------------------

# Monthly Profit Trend

### Objective
 Monitor profitability over time.

### Visualization
 Line Chart

# Monthly Profit Trend

```python
monthly_profit = (
    df.groupby(["year","month"])["profit"]
      .sum()
      .reset_index()
)

plt.figure(figsize=(12,5))

plt.plot(monthly_profit["profit"])

plt.xlabel("Timeline (Sequential Months)", fontsize=11)

plt.ylabel("Total Profit ($)", fontsize=11)

plt.show()
```

## Observations
- Profit changes throughout the year.
- Some months achieve higher profitability despite similar sales.
- Profit trends help evaluate operational performance.

# Correlation Analysis

### Objective
 Measure relationships between numerical variables.

### Visualization
 Correlation Heatmap

# Correlation Analysis

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

## Observations
- Strong positive correlations indicate variables that increase together.
- Weak correlations suggest limited relationships.
- Correlation analysis supports feature selection for predictive models.
